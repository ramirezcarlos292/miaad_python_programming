# Objetivo
Utilizar GridSpec para organizar subplots de diferentes tamaños en un análisis de ventas.

# Instrucciones:

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
```

```python
'''
1. Genera un DataFrame de ventas con Mes, Ventas, y Gastos. Utiliza randint para
generar números aleatorios enteros. Para ventas los valores deben estar entre
2000 y 5000. Para gastos, los valores debn estar entre 1000 y 4000.
'''

ventas = np.random.randint(2000, 5000, size=12)
gastos = np.random.randint(1000, 4000, size=12)
mes = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic']

df = pd.DataFrame({'mes': mes, 'ventas': ventas, 'gastos': gastos})
df
```

```python
'''
2. Usa GridSpec para organizar los subplots de la siguiente manera:
  - Primer gráfico (barra, ocupando dos columnas).
  - Segundo gráfico (línea, una columna).
  - Tercer gráfico (dispersión, dos filas).
'''

fig = plt.figure(figsize=(10, 7))
fig.suptitle('Reporte Anual de Ventas y Gastos')

gs = GridSpec(3, 2, left=0.05, right=1, wspace=0.15, hspace=0.5)
ax1 = fig.add_subplot(gs[0, :])
a_dict = {'ventas':ventas, 'gastos': gastos}
bottom = np.zeros(12)
for mon, data in a_dict.items():
  p = ax1.bar(df['mes'], height=data, label=mon, bottom=bottom)
  ax1.legend(loc='upper right', ncols=2)
  bottom += data

ax2 = fig.add_subplot(gs[1, :-1])
ax2.plot(mes, df['ventas'], label='ventas')
ax2.plot(mes, df['gastos'], label='gastos')
ax2.legend(loc='upper right')

ax3 = fig.add_subplot(gs[2, :])
ax3.scatter(mes, df['ventas'], label='ventas')
ax3.scatter(mes, df['gastos'], label='gastos')
ax3.legend(loc='upper right')
```
