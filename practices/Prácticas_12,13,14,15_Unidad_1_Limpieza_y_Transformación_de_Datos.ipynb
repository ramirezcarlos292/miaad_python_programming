# Práctica 11: Carga de Datos

Al final del archivo contiene una línea que dice "Esta es una línea que no deberías cargar". Esta línea es irrelevante por lo tanto al cargar el CSV a un dataframe se debería de ignorar.

- Carga el archivo a un dataframe usando el código necesario para omitir la última fila.
- Usa df.tail(3) para demostrar que no has cargado la última línea
- Imprime la cantidad de filas y columnas del dataframe

```python
import pandas as pd
import numpy as np

# Omitir ultima fila usando skipfooter
df = pd.read_csv('/content/drive/MyDrive/ClassFiles/miaad-nyc-r-s.csv', skipfooter=1)

# Usa df.tail(3) para demostrar que no has cargado la última línea
df.tail(3)
```

```python
# Imprime la cantidad de filas y columnas del dataframe
rows, columns = df.shape
print(f"Número de Filas: {rows}, Número de Columnas: {columns}")
```

# Práctica 12: Agregar una columna

- Imprime los nombres de la columnas del archivo
- Asegúrate que el nombre de la primera columna debería ser INDEX MIIAD. Si esto no es así, entonces agrega dicho nombre a la columna y vuelve a imprimir los nombres para mostrar el cambio realizado.

```python
# Imprimir los nombres de columnas del archivo
columns_list = list(df.columns)
print(columns_list)

# Renobrar primer columna
columns_list_new = columns_list.copy()
columns_list_new[0] = 'INDEX_MIAAD'

df = df.rename(columns=dict(zip(columns_list, columns_list_new)))

# Imprimir los nombres de columnas para observar el cambio
print(list(df.columns))
```

# Práctica 13: Normalización de los nombres de la columna.

- Normaliza los nombres de ls columnas, los cuales deberán estar en minúsculas. Asimismo, reemplaza los espacios con guion bajo
- Imprime los nombres de las columnas para demostrar que has realizado la normalización.

```python
# Normaliza los nombres de ls columnas, los cuales deberán estar en minúsculas. Asimismo, reemplaza los espacios con guion bajo.
columns_list = list(df.columns)
print(columns_list)

# Renobrar primer columna
columns_list_new = list(df.columns.str.lower().str.replace(' ', '_'))

df = df.rename(columns=dict(zip(columns_list, columns_list_new)))

# Imprimir los nombres de columnas para observar el cambio
print(list(df.columns))
```

# Práctica 14: Errores en los Tipos de Datos

- Imprime los tipos de datos de cada columna.
- Asegúrate que las columnas de SALE PRICE(sale_price), LAND SQUARE FEET (land_square_feet), GROSS SQUARE FEET (gross_square_feet) sean de tipo flotante. Escribe qué tipo de dato tienen asignado. Si no son correctos, realiza el cambio.
- Imprime de nuevo los tipos de datos para demostrar que has arreglado el problema.

```python
# Imprime los tipos de datos de cada columna.
column_types = df.dtypes.reset_index()

# Asegúrate que las columnas de SALE PRICE(sale_price),
# LAND SQUARE FEET (land_square_feet),
# GROSS SQUARE FEET (gross_square_feet) sean de tipo flotante.

# Escribe qué tipo de dato tienen asignado. Si no son correctos, realiza el cambio.
print(f"sale_price data type is: {column_types.loc[column_types['index'] == 'sale_price'][0]}")
print(f"land_square_feet data type is: {column_types.loc[column_types['index'] == 'land_square_feet'][0]}")
print(f"gross_square_feet data type is: {column_types.loc[column_types['index'] == 'gross_square_feet'][0]}")
```

```python
# Como los tipos de datos para estas columnas deberian ser flotantes, realizamos el cambio
df['sale_price'] = pd.to_numeric(df['sale_price'], errors='coerce')
df['land_square_feet'] = pd.to_numeric(df['land_square_feet'], errors='coerce')
df['gross_square_feet'] = pd.to_numeric(df['gross_square_feet'], errors='coerce')

# Imprime de nuevo los tipos de datos para demostrar que has arreglado el problema.
column_types = df.dtypes.reset_index()
print(f"new sale_price data type is: {column_types.loc[column_types['index'] == 'sale_price'][0]}")
print(f"new land_square_feet data type is: {column_types.loc[column_types['index'] == 'land_square_feet'][0]}")
print(f"new gross_square_feet data type is: {column_types.loc[column_types['index'] == 'gross_square_feet'][0]}")
```

# Práctica 15:
Genera el diccionario de datos del dataframe preprocesado. Imprime el contenido de dicho diccionario

```python
diccionario_datos = pd.DataFrame({'Nombre de columna': df.columns,
                                  "Tipo de dato": [str(df[col].dtype) for col in df.columns]})
print(diccionario_datos)

diccionario_datos.to_csv('/content/drive/MyDrive/ClassFiles/diccionario_datos.csv', index=False, encoding='utf-8')
```
