# **Ejercicio 01:** Errores Tipográficos y Valores Inconsistentes: maritalstatus
* Imprime de la columna 'maritalstatus' los valores únicos.
* Imprime la frecuencia de cada valor único
* Cambia el valor 'Never-Married' a 'Single'
* Imprime la frecuencia de cada valor único.

```python
print(f"Valores únicos de la columna maritalstatus")
maritalstatus = list(df.maritalstatus.unique())
for item in df.maritalstatus.unique():
  print(f"- {item}")
```

```python
print(f"Frecuencia de cada valor único")
msdf = df.maritalstatus.value_counts().reset_index()
msdf
```

```python
null_ms = df.maritalstatus.isna().value_counts().reset_index()
null_maritalstatus = null_ms['count'].loc[null_ms['maritalstatus'] == True]
print(f"Hay {null_maritalstatus.values[0]} datos NaN in la columna 'maritalstatus'")
nan_ms = pd.DataFrame({'maritalstatus': ['Not reported'], 'count': null_maritalstatus.values[0]})
```

```python
# Cambia el valor 'Never-Married' a 'Single'
df.loc[df['maritalstatus'] == 'Never-married', 'maritalstatus'] = 'Single'
```

```python
print(f"Imprime la frecuencia de cada valor único")
msdf = df.maritalstatus.value_counts().reset_index()
msdf = pd.concat([msdf, nan_ms])
msdf.sort_values('count', ascending=False)
```

# **Ejercicio 02:** Errores Tipográficos y Valores Inconsistentes: highestdegree.
* Imprime de la columna 'highestdegree' los valores únicos.
* Imprime la frecuencia de cada valor único
* Observa que cada valor categórico empieza con 0. Modifica las categorías para que todas estén en minúsculas y ninguna empiece con un número. Por ejemplo, si dice "2. High School", debería ser 'high school.
* Los cambios se realizarán en la misma columna 'highestdegree'.
* Imprime los nuevos valores únicos.

```python
print(f"Valores únicos de la columna 'highestdegree")

for item in list(df['highestdegree'].unique()):
  print(f"- {item}")
```

```python
print(f"Frecuencia de cada valor único")
df['highestdegree'].value_counts().reset_index()
```

```python
print(f"Frecuencia de cada valor único después de corregir columna 'highestdegree'")
df['highestdegree'] = df['highestdegree'].str.lower().str.replace(r'\d+\.' , '', regex=True)
df['highestdegree'].value_counts()
```

# **Ejercicio 03:** Alta Cardinalidad - Frecuencias Bajas
* De la columna 'maritalstatus' une las categorías Widowed y 'Single' en una categoría más general llamada 'Single/No Partner'. Crea una nueva columna, 'marital_grouped' donde aparezcan las nuevas categorías de maritalstatus.
* Imprime los valores únicos de la nueva columna 'marital_grouped'.
* Imprime la frecuencia de los valores únicos de la nueva columna 'marital_grouped'.

```python
df['marital_grouped'] = df['maritalstatus'].apply(lambda x: 'Single/No Partner' if x == 'Widowed' else ('Single/No Partner' if x == 'Single' else x))

print("Nuevos valores únicos después de unir Widowed y Single")
for item in list(df['marital_grouped'].unique()):
  print(f"- {item}")
```

```python
print(f"Frecuencia de los valores únicos de la nueva columna 'marital_grouped'")
df['marital_grouped'].value_counts()
```
