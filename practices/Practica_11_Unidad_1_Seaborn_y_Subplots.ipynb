# Objetivo
Utilizar Seaborn junto con plt.subplots() para visualizar diferentes aspectos de un conjunto de datos en una sola figura.

# Instrucciones:
1. Carga el dataset "penguins" de Seaborn.
  - Utiliza df = sns.load_dataset("penguins")
  - Usa plt.subplots() para crear una figura con 2 filas y 2 columnas.
  - Genera los siguientes gráficos:
    * (0,0): sns.histplot(): Histograma de las longitudes de aletas.
    * (0,1): sns.boxplot(): Boxplot comparando la masa corporal por especie.
    * (1,0): sns.scatterplot(): Dispersión entre masa corporal y largo de aletas.
    * (1,1): sns.violinplot(): Violin plot de la masa corporal según la especie.
  - Personaliza los gráficos con títulos y colores adecuados.
  - Ajusta el diseño con plt.tight_layout().

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset("penguins")
df.head()
```

```python
fig, axes = plt.subplots(2, 2, sharex=False, sharey=False, figsize=(10, 6))

# histograma
sns.histplot(df, x='flipper_length_mm', ax=axes[0, 0], hue='species')
axes[0, 0].set_ylabel('Observaciones')
axes[0, 0].set_xlabel('Longitud de Aleta (mm)')
axes[0, 0].set_title('Distribución de tamaño de aletas')
# box plot
sns.boxplot(df, x='species', y ='body_mass_g', ax=axes[0, 1], hue='species')
axes[0, 1].set_ylabel('Masa corporal (gr)')
axes[0, 1].set_xlabel('Especie')
axes[0, 1].set_title('Masa corporal por especie')
# Scatter plot
sns.scatterplot(df, x='body_mass_g', y='flipper_length_mm', ax=axes[1, 0], hue='species')
axes[1, 0].set_ylabel('Lon de aleta (mm)')
axes[1, 0].set_xlabel('Masa corporal (gr)')
axes[1, 0].set_title('Dispersión masa corporal y lon. de aleta')
axes[1, 0].legend(loc='lower right')
# violin plot
sns.violinplot(df, x='species', y ='body_mass_g', ax=axes[1, 1], hue='species')
axes[1, 1].set_ylabel('Masa corporal (gr)')
axes[1, 1].set_xlabel('Especie')
axes[1, 1].set_title('Masa corporal por especie')

fig.suptitle('Explorando dataset de pinguinos')
plt.tight_layout()
plt.show()
```
