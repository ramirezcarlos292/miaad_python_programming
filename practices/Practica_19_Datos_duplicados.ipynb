### Ejercicio 01
Duplicidad Parcial Una columna: Contar y Filtrar los registros duplicados en Duration

**1.1** Carga el conjunto de datos dirty.csv y verifica que su carga sea correcta. <br>
**1.2** Cuenta la cantidad de registros duplicados utilizando solo la información de columna Duration e imprime.<br>
**1.3** Utiliza value_counts() e imprime la cantidad de registros que hay por cada valor único.<br>
**1.4** Filtra los datos y obten los registros sin duplicados.<br>

```python
print(f"1.1 Carga el conjunto de datos:")
df = pd.read_csv('/content/drive/MyDrive/ClassFiles/dirtydata.csv')
df.head(5)
```

```python
print(f"1.2 Cuantos datos duplicados existen en la columna 'Duration'?")
print(f"    En la columna 'Duration' existen: {int(df.duplicated(subset=['Duration']).sum())} datos duplicados")
```

```python
print(f"1.3 Utiliza value_counts() e imprime la cantidad de registros que hay por cada valor único.")
print(f"    Cuenta de valores únicos en la columna Duration")
df['Duration'].value_counts()
```

```python
print(f"1.4 Filtra los datos y obten los registros sin duplicados.")
df.drop_duplicates(subset=['Duration'], keep=False, inplace=True)
print(len(df))
df
```

### Ejercicio 02
Duplicidad Parcial Dos columna: Contar y Filtrar los registros duplicados en Pulse y MaxPulse

**2.1** Carga el conjunto de datos dirty.csv y verifica que su carga sea correcta.<br>
**2.2** Cuenta la cantidad de registros duplicados utilizando la información de las columnas Pulse y Maxpulse.<br>
**2.3** Utiliza value_counts() e imprime la cantidad de registros que hay por cada valor único.<br>
**2.4** Filtra los datos y obten los registros cuando keep = 'last'.<br>

```python
print(f"2.1 Carga el conjunto de datos:")
df = pd.read_csv('/content/drive/MyDrive/ClassFiles/dirtydata.csv')
df_original_len = len(df)
print(f"Cantidad de rows del dataset original {df_original_len}")
df.head(5)
```

```python
print(f"2.2 Cuantos datos duplicados existen en la columna 'Pulse' y 'Maxpulse'?")
print(f"    En las columns 'Pulse' existen: {int(df.duplicated(subset=['Pulse'], keep=False).sum())} datos duplicados")
print(f"    En las columns 'Maxpulse' existen: {int(df.duplicated(subset=['Maxpulse'], keep=False).sum())} datos duplicados")
```

```python
print(f"2.3 Utiliza value_counts() e imprime la cantidad de registros que hay por cada valor único.\n")
print(f"----Cuenta de valores únicos en la columna 'Pulse': \n{df['Pulse'].value_counts()}\n")
print(f"----Cuenta de valores únicos en la columna 'Maxpulse':\n{df['Maxpulse'].value_counts()}\n")
```

```python
print(f"2.4 Filtra los datos y obten los registros cuando keep = 'last'.")
df.drop_duplicates(subset=['Pulse', 'Maxpulse'], keep='last', inplace=True)
print(f"Después de remover duplicados de columnas, 'Pulse' y 'Maxpulse', quedaron y conservar 'last': {len(df)}")
df
```
